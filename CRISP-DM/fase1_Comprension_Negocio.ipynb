# Fase 1 - Comprensión del negocio

Proyecto: **predicción de suscripción a depósito a plazo (Bank Marketing)**, metodología CRISP-DM.

## 1.1 Contexto del problema
Una entidad bancaria realiza campañas de marketing directo para ofrecer un depósito a plazo. Cada contacto tiene un costo y solo una minoría de los clientes contratados acepta la oferta. Estimar la propensión de cada cliente a suscribir permite priorizar la gestión comercial y usar mejor los recursos de la campaña.

## 1.2 Objetivo general
Construir un modelo de **clasificación binaria** que prediga si un cliente suscribirá un depósito a plazo (`y` = `yes` / `no`) y entregue una probabilidad que apoye la priorización de contactos.

## 1.3 Objetivos específicos
1. Comprender la estructura, calidad y desbalance del dataset `bank-full.csv`.
2. Preparar los datos evitando fuga de información (partición train / validación / test ciego).
3. Desarrollar **2 iteraciones** de modelado con **3 algoritmos** cada una (Regresión Logística, Árbol de Decisión y Random Forest).
4. Comparar iteraciones con métricas adecuadas a un problema desbalanceado y evaluar la generalización con datos no vistos.
5. Desplegar una aplicación Streamlit que permita seleccionar variables, consultar los modelos en tiempo real y comparar su desempeño.

## 1.4 Criterios de éxito
| Criterio | Meta de referencia |
|---|---|
| Recall de la clase `yes` (test ciego) | ≥ 0,50 |
| ROC-AUC (test ciego) | ≥ 0,70 |
| Diferencia validación vs. test | pequeña (sin sobreajuste evidente) |
| Mejora Iteración 2 vs. Iteración 1 | mejora en F1 o Recall con ROC-AUC similar o mayor |

Por el desbalance de clases, **Accuracy no se usa como métrica principal**: se priorizan Recall, F1, ROC-AUC y PR-AUC. Los criterios se verifican de forma explícita en la Fase 5.

## 1.5 Supuestos y restricciones
- El conjunto de datos es representativo de campañas futuras similares.
- Se excluyen `duration`, `campaign`, `day`, `month` y `contact`: describen el contacto actual y no estarían disponibles al priorizar clientes antes de llamar (evita fuga de información).
- Los valores `unknown` se conservan como una categoría explícita.
- El test ciego (20 %) no se usa para seleccionar variables, modelos ni hiperparámetros.
- Herramientas: Python, pandas, scikit-learn, matplotlib, joblib y Streamlit.

## 1.6 Plan de iteraciones (mínimo 2)
| Iteración | Descripción | Algoritmos |
|---|---|---|
| 1 | Modelo base con preprocesamiento estándar, sin balanceo ni selección avanzada | Regresión Logística, Árbol de Decisión, Random Forest |
| 2 | Selección de características (`SelectPercentile`), balanceo con pesos de clase y ajuste con `GridSearchCV` | Regresión Logística, Árbol de Decisión, Random Forest |

Fases siguientes: Fase 2 (datos) → Fase 3 (preparación) → Fase 4 (modelado) → Fase 5 (evaluación) → Fase 6 (despliegue).
